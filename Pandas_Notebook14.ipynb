## Pandas Dataframe

#### Here Dictionary is the foundation of a dataframe

```python
heights = {"John": 1.70, "Alice": 1.65, "Fred": 1.78, "Jane": 1.50}

weights = {"John": 60, "Alice": 50, "Fred": 70, "Jane": 55}
```

```python
#Don't use "print" to print your DataFrame because it doesn't format it properly the way you want it

bmi_info = pd.DataFrame({"height": heights, "weight": weights})

bmi_info
```

```python
type(bmi_info)
```

```python
#Anothe way to create a DataFrame

data = {"Name": ["John", "Alice", "Fred", "Jane"], "Height": [1.70, 1.65, 1.78, 1.50], "Weight": [60, 50, 70, 55]}
```

```python
df = pd.DataFrame(data)
```

## Different ways to access data in a DataFrame

#### Using loc

```python
Clean_bmi_info
```

```python
Clean_bmi_info
```

```python
Clean_bmi_info.index
```

```python
#Use loc to select an element from the DataFrame

Clean_bmi_info.loc["John", "height"]
```

```python
Clean_bmi_info.loc["Fred", "weight"]
```

```python
#Slicing

Clean_bmi_info.loc[:"Fred", :]
```
