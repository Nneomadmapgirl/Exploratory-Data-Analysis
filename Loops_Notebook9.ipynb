## for Loops

**for loops are used with iterables(strings, tuples, sets, dictionaries, string)**

```python
#**using loops with string**
greeting = "Hello"

print(greeting)
```

```python
#in the for loop, you have a placeholder and you name it anything you feel like because its just used to represent each xter/item
#char serves as the placeholder
for char in greeting:
    print(char)
```

```python
for char in greeting:
    print(f"The letter {char} is present in greeting")
```

**Using the for loop with list**

```python
student_names = ["John", "Alice", "Fred"]

print(student_names)
```

```python
for huot in student_names:
    print(f"Hi, {huot}")
print("Welcome to Techcrush")
```

```python
#this is to show you the work of indentation, notice how the print got indented

for huot in student_names:
    print(f"Hi, {huot}")
    print("Welcome to Techcrush")
```

```python
#when you use the enumerate with list fxn, it gives you the index/values pair of the item in a list in a tuple
#enumerate is a generator object so it doesn't work alone until you put it into a data structure

list(enumerate(student_names))
```

```python
indexed_list = list(enumerate(student_names))
```

```python
print(indexed_list)
```

```python
for ye, ta in indexed_list:
    print(f"The index {ye} and name with the index is {ta}")
```

```python
for index, name in indexed_list:
    index = index + 1 #this just adds one to any index it fetches
    print(f"The number {index} student is {name}")
```

**using loop with dictionary**

```python
favourite_fruits = {"Alice":"Pear", "Fred":"Apple", "John":"Grape"}

print(favourite_fruits)
```

```python
print(f"The names in the dictionary are:")
for key in favourite_fruits.keys():
    print(key)
```

```python
for key, value in favourite_fruits.items():
    print(f"{key}'s favourite fruit is {value}")
```

**Range Function** 

```python
#list of numbers from 1 to 10

list(range(1, 11))
```

```python
list(range(10, 21))
```

```python
list(range(10, 21, 2)) #given it a step of two
```

```python
for num in range(1, 11):
    print(num)
```

```python
#to create a list of even numbers from 0 to 20

even_numbers = []
for num in range(0, 21):
    if num % 2 == 0:
        even_numbers.append(num)
```

```python
print(even_numbers)
```

```python
#to create a list of odd numbers from 0 to 20

odd_numbers = []
for num in range(0, 21):
    if num % 2 == 1:
        odd_numbers.append(num)
```

```python
print(odd_numbers)
```

```python
#to create a list of odd numbers from 0 to 20

even_numbers = []
odd_numbers = []
for num in range(0, 21):
    if num % 2 == 0:
        even_numbers.append(num)
    elif num % 2 == 1:
        odd_numbers.append(num)
```

```python
print(odd_numbers)
```

```python
#to create a list of squares from 1 to 10

squares = []

for num in range(1, 11):
    square = num ** 2
    squares.append(square)
    
```

```python
print(squares)
```

## List Comprehension

```python
square_comprehension = [num**2 for num in range(1, 11)]

print(square_comprehension)
```

```python
even_numbers_comp = [num for num in range(1, 21) if num % 2 == 0]

print(even_numbers_comp)
```

```python
odd_numbers_comp = [num for num in range(1, 21) if num % 2 == 1]

print(odd_numbers_comp)
```
