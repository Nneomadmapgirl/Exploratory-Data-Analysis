## Joining Datasets in Pandas

```python
import pandas as pd

import numpy as np
```

## Using Concat

```python
df1 = pd.DataFrame({"A":[1,2], "B":[3,4]})

df1
```

```python
df2 = pd.DataFrame({"C":[5,6], "D":[7,8]})

df2
```

```python
#To concatenate both frames now

df3 = pd.concat([df1,df2], axis = 1) #The axis= 1 will concatenate column_wise

df3
```

```python
df4 = pd.concat([df1,df2], axis = 0) #To concatenate along rows

df4
```

```python
df4.isna().sum()
```

```python
df4.fillna(0, inplace = True)
```

```python
df4
```

## Using Merge(There are different types of join here)

```python
df5 = pd.DataFrame({"ID":[1,2,3], "Name":["Alice", "Bob", "Charles"]})

df5
```

```python
df6 = pd.DataFrame({"ID":[2,3,4], "Score":[85, 90, 95]})

df6
```

### Inner Join

**This returns records present in both tables based on the common column**

```python
pd.merge(df5,df6, on = "ID", how = "inner")
```

### The Left Join

**This returns all the records present in the left table and record present in both tables**

```python
pd.merge(df5,df6, on = "ID", how = "left")#Remember the left table is the table you first specify in your arguement
```

## The Right Join

**This returns all the records present in the right table and records present in both tables**

```python
pd.merge(df5,df6, on = "ID", how = "right")
```

```python
df7 = pd.DataFrame({"Student_ID":[2,3,4], "Score":[85, 90, 95]})

df7
```

```python
#for instance if the column are named differently in the data frame and you don't want to rename and that's the common column you want to join on

df8 = pd.merge(df5, df7, left_on = "ID", right_on = "Student_ID", how = "inner")

df8
```

```python
df8.drop("Student_ID", axis = 1, inplace = True)
```

```python
df8
```

## Outer Join

**This returns all colums(all records)**

```python
pd.merge(df5,df6, on = "ID", how = "outer")
```
