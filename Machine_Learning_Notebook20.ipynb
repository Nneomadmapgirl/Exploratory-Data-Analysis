```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

#Anything you do in supervised machine learning, you  it from scikit-library

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
df = pd.read_csv(r"C:\Users\USER\Downloads\TECHCRUSH_Datasets(Techcrush)\train (1).csv")

df.head()
```

## Step 1: Data Cleaning

```python
df.info()
```

```python
df.shape
```

```python
df.isna().sum() #null values
```

```python
#Check the percentage of null values in each column

missing_percent = df.isna().mean() * 100

missing_percent[missing_percent > 0]
```

**Once its greater than 30%, its not very useful and should be dropped**

```python
cols_to_drop = missing_percent[missing_percent > 30].index #The .index is to get the column names

cols_to_drop
```

```python
df.drop(columns = cols_to_drop, inplace = True)
```

```python
missing_percent = df.isna().mean() * 100

missing_percent[missing_percent > 0]
```

**Filling of null values**

```python
sns.boxplot(df["LotFrontage"])#just used the lotfootage to confirm there are outliers in all, the lotfootage is just a sample

plt.show()
```

```python
df.info()
```

```python
#Filtering to numerical columns

numerical_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

numerical_cols
```

```python
#using for loop to fill null values in numerical columns

for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
missing_percent = df.isna().mean() * 100

missing_percent[missing_percent > 0]
```

```python
#Filtering to categorical columns

categorical_cols = df.select_dtypes(include=["object"]).columns.tolist()

categorical_cols
```

```python
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
missing_percent = df.isna().mean() * 100

missing_percent[missing_percent > 0]
```

```python
df.info()
```

## Feature Encoding

```python
df.head()
```

**We are taking the first type which is a supervised learning(regression)**

```python
#x is the independent variable
#y is the dependent variable
```

```python
X = df.drop(columns=["SalePrice","Id"])

y = df["SalePrice"]  #y is what we are predicting which is our sales price
```

```python
X
```

**We have to encode these categorical variables to numbers because machine learning only deals with numbers. That's the essence of feature encoding**

```python
categorical_cols
```

## Label Encoder

```python
X_lab = X.copy()
```

```python
for col in categorical_cols:
    le = LabelEncoder() #create an instance and name it anything of your choice, i named mine le
    X_lab[col] = le.fit_transform(X_lab[col])
```

```python
X_lab.head()
```

```python
X.head()
```

## One Hot Encoder

```python
X_oh = X.copy()
```

```python
X_oh = pd.get_dummies(X_oh, columns = categorical_cols, dtype = "int")
```

```python
X_oh.head()
```

```python
X["MSZoning"].value_counts()
```

```python
X_oh.shape
```

```python
X.shape
```

## Feature Scaling
**This is adjusting the range of our data to a common ground**

```python
X_lab.head()
```

**Using Standard Scaler: mean = 0, std = 1**

```python
scaler = StandardScaler()

Scaledx = scaler.fit_transform(X_lab)
```

```python
Scaledx
```

## Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(Scaledx, y, test_size = 0.2, random_state = 42) #80:20 split 
```

```python
X_train.shape
```

```python
X_test.shape
```

## Training

```python
linreg = LinearRegression()

linreg.fit(X_train, y_train) #fit the model to the training set
```

```python
lin_preds = linreg.predict(X_test) #we are predicting

lin_preds
```

```python
y_test
```

## Evaluation Metrics
**The model with the lowest error is the best model**

**The major goal for any regression task is to reduce the errors**

```python
r2_score(y_test, lin_preds)
```

```python
mean_absolute_error(y_test, lin_preds)
```

```python
mean_squared_error(y_test, lin_preds)
```

```python
np.sqrt(mean_squared_error(y_test, lin_preds))
```
