## Aggregating and Grouping

```python
sales_data.head()
```

```python
#To see the countries the values are located in

sales_data["Country"].value_counts()
```

```python
#To see the states the values are located in

sales_data["State"].value_counts()
```

```python
#For example you want to find the total sales in California, first, filter the data to only California

total_sales_california = sales_data[sales_data["State"] == "California"]["Sales"].sum()

total_sales_california
```

```python
#If you want to find the total sales for each state

total_sale_by_state = sales_data.groupby("State")["Sales"].sum().reset_index()

total_sale_by_state
```

```python
#Lets say you want to find the state with the higest total sales

total_sale_by_state.sort_values(by = "Sales", ascending = False, inplace = True)
```

```python
total_sale_by_state
```

```python
#If you just want to work with the first five

total_sale_by_state[:5]
```

```python
#If you want to find average sales

avg_sales_by_state = sales_data.groupby("State")["Sales"].mean().reset_index()

avg_sales_by_state.sort_values(by = "Sales", ascending = False, inplace = True)

avg_sales_by_state
```

```python
#Calculating the avrage state_sales all at once

avg_total_sales = sales_data.groupby("State")["Sales"].agg([np.sum, np.mean]).reset_index()

avg_total_sales
```

```python
sales_data.head()
```

```python
sales_category_sub = sales_data.groupby(["Category", "Sub-Category"])["Sales"].agg([np.sum, np.mean])

sales_category_sub
```

## Pivoting

```python
sales_by_category = sales_data.pivot_table(index = "Category", columns = "Sub-Category", values = "Sales", aggfunc=np.sum)

sales_by_category 
```

```python
#To replace the Not a Number(NaN) by zero, add the last fill value = 0

sales_by_category = sales_data.pivot_table(index = "Category", columns = "Sub-Category", values = "Sales", aggfunc=np.sum, fill_value = 0)

sales_by_category 
```
