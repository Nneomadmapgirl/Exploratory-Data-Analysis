## To correct errors, use the try except block

```python
first_number = float(input("Enter the first number: "))
second_number = float(input("Enter the second number: "))
result = first_number/second_number
print(result)
```

```python
#To correct the above, use the try_except block

try:
    first_number = float(input("Enter the first number: "))
    second_number = float(input("Enter the second number: "))
    result = first_number/second_number
except ZeroDivisionError:
    print("You can't divide by zero")
else:
    print(result) #you will get the result if you enter valid numbers
    
```

```python
#To correct the above, use the try_except block

try:
    first_number = float(input("Enter the first number: "))
    second_number = float(input("Enter the second number: "))
    result = first_number/second_number
except ValueError:
    print("You didn't enter a valid number, enter a valid number")
else:
    print(result)
    
```

```python
#Your try_except block can also have a finally block

#To correct the above, use the try_except block

try:
    first_number = float(input("Enter the first number: "))
    second_number = float(input("Enter the second number: "))
    result = first_number/second_number
except ZeroDivisionError:
    print("You can't divide by zero")
else:
    print(result)
    
finally:
    print("Thank you for using our programme")
```

```python
#To handle all errors once instead of specifying each error

try:
    first_number = float(input("Enter the first number: "))
    second_number = float(input("Enter the second number: "))
    result = first_number/second_number
except:
    print("Your input is not valid")
```
