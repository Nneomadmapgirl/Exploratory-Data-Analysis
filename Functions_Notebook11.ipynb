## User Defined Fxns

**Sometimes as a programmer, you have to create your own functions**

```python
#Let's define fxns without parameters first

def greet_user():
    print("Hello, everyone")
```

```python
#To call the fxn

greet_user()
```

```python
#To define a fxn with parameters

def greet_user_parameter(name):
    print(f"Hello,{name}")
    
```

```python
greet_user_parameter("Anne")
```

```python
greet_user_parameter(name)
```

```python
def sum_two_numbers(a, b):
    result = a + b
    print(result)
```

```python
sum_two_numbers(2, 3)
```

```python
#result of a variable cannot be in a variable using print, it will give you none

first_result = sum_two_numbers(2, 3)
```

```python
print(first_result)
```

```python
#It is always good to use return

def sum_two_numbers(a, b):
    result = a + b #local variable, can only be accessed within the fxn
    return result
```

```python
print(result)
```

Any variable defined in a function is a local variable, meaning it can't operate outside the function
such as the result in the above cell

```python
second_result = sum_two_numbers(4, 3) #global variable
```

```python
print(second_result)
```

## Positional Arguments

```python
def describe_student(name, course):
    return f"My name is {name}, and I am taking {course} course"
```

```python
describe_student("Justice", "Python")
```

```python
#This is why beign careful of input arrangement is necessary

describe_student("Python", "Justice")
```

## Keyword Arguments 

```python
describe_student(course = "Python", name = "Justice")
```

## Default Values

```python
def describe_student(name, course = "Python", age = "18"):
    return f"My name is {name}, I am {age} years old and I am taking {course} course on Techcrush"
```

```python
describe_student("Fred")
```

```python
#if you want to overwrite default values

describe_student("Fred", "SQL")
```

 ## Passing a list

```python
def list_sum(num_list):
    result = sum(num_list)
    return result
```

```python
list_sum([1,2,3])
```

## Passing an arbitrary numer of argument

```python
def sum_of_numbers(*numbers):
    result = 0
    for num in numbers:
        result += num
    return result
```

```python
sum_of_numbers(1,2,3,4,5,6)
```

## Passing an arbitrary numer of keyword argument

```python
def student_profile(**student_info):
    return student_info
```

```python
student_dict = student_profile(name="Ify", course="SQL", gender="female")
```

```python
print(student_dict)
```

```python
student_dict.keys()
```

```python
student_dict.values()
```

## Anonymous Functions

```python
#To add two variables

add = lambda x, y:x+y
```

```python
print (add(2, 3))
```

```python
power = lambda a, b: a**b
```

```python
print(power(4, 5))
```

#### Filter Function

```python
#using with the filter function

numbers = [1, 2, 3, 4, 5, 6]
```

```python
#the x is representing each number in the list, you can use any alphabet you choose

even_numbers = filter(lambda x: x % 2 == 0, numbers) 
```

```python
print (list(even_numbers))
```

```python
odd_numbers = filter(lambda x: x % 2 == 1, numbers)
```

```python
print (list(odd_numbers))
```

#### Using with the map function

```python
print(numbers)
```

```python
squares = map(lambda h: h ** 2, numbers)
```

```python
print(list(squares))
```
