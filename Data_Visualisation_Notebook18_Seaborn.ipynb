## There are two elements in a matplotlib

-The first element is a figure: It is the entire canvas that holds all subplots

-The second is the axes: It holds a single plot

```python
#To create a figure with a single axis(meaning a single plot)

x = [1, 2, 3]

y = [4, 5, 6]

#To initialise a plot, we the fig and the ax which serve as place holders

fig, ax = plt.subplots()

ax.plot(x,y)

ax.set_title("Basic_Plot")
ax.set_xlabel("X_Axis")
ax.set_ylabel("Y_Axis")

#To see your plot, always ensure you use plt.show

plt.show()
```

```python
#To create a figure with multiple subplots

fig, axes = plt.subplots(1, 2)

axes[0].plot(x, y)
axes[1].plot(x,y)
axes[1].set_title("Right plot")
axes[0].set_title("Left plot")

fig.subplots_adjust(right=0.9)

plt.show()
```

```python
#To create a figure with multiple subplots

fig, axes = plt.subplots(2, 2)

axes[0,1].plot(x, y)
axes[0,1].plot(x, y)
axes[1,0].plot(x, y)
axes[1, 1].plot(x, y)
               
plt.show()
```

```python
x = rating_count.index.astype(str)
y = rating_count.values
colors = plt.cm.plasma(np.linspace(0, 1, len(y)))

plt.figure(figsize=(8,5))
bars = plt.bar(x, y, color=colors)

# Add labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,  # center text horizontally
        height,                          # position at top of bar
        f'{height:.0f}',                  # value (no decimals)
        ha='center', va='bottom'          # align text
    )

plt.title("Rating Counts")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

## Bivariate Analysis

```python
#Movies and TV shows Release

release_year_by_type = df.groupby(["release_year", "type"]).size().unstack(fill_value = 0).reset_index(drop=False)
                                                                                                    
release_year_by_type
```

```python
#We are going to use a line plot because we are dealing with numerical data
#We will fetch everything from the dataframe

plt.plot(release_year_by_type["release_year"], release_year_by_type["Movie"], label ="Movie", color = "green")

plt.plot(release_year_by_type["release_year"], release_year_by_type["TV Show"], label ="TV Show", color = "orange")

plt.legend()

plt.title("Movies and TV Shows Released Over the years")

plt.show()
```

```python
df.head()
```

```python
df["duration"].value_counts()
```

```python
#We need this duration column to be a numerical data column so we need to clean the seasons and minutes before visualizing
#To extract the numerical part in a string we use regex

df["duration_mins"] = df["duration"].str.extract("(\d+)").astype("float")
```

```python
df["duration_mins"].value_counts()
```

```python
#Now some of these duration were not actually in minutes, so we need to convert them to minutes, we can make assumptions by using the average minutes in a season to work with all

df["duration_mins"] = df["duration_mins"].replace({1.0:500, 2.0:1000, 3.0:1500})
```

```python
df["duration_mins"].value_counts()
```

```python
duration_by_year = df.groupby("release_year")["duration_mins"].mean().reset_index()

duration_by_year
```

```python
plt.scatter(duration_by_year["release_year"], duration_by_year["duration_mins"], color = "green")

plt.show()#The scatter plot is just showing the relationship
```

**From the above chart, we can see it was a gradual increase from 1940 to 2000 and then a peak rise around 2020**

## Boxplot is also used for distribution of numerical variables

## Boxplot is for univariate while scatter plot is for bivariate

```python
sns.boxplot(data = df, x="duration_mins")

plt.show()
```

-The line at the middle is the median and median is also called 50th percentile

-The line at the left side of the box is called 25th percentile(Q1)

-The line at the right side of the box is the 75th percentile(Q3)

-IQR = Q3-Q1

-Whisker at the left-hand side is 1.5*IQR from Q1

-Whisker at the right-hand side is 1.5*IQR from Q3

-Any data point outside the whisker is an outlier
